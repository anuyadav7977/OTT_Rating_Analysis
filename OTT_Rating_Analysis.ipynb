```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/amazon_prime_movies.csv")



# Convert rating into numbers
df["IMDb Rating"] = pd.to_numeric(
    df["IMDb Rating"], errors="coerce"
)

# Average rating
print("Average IMDb Rating:", df["IMDb Rating"].mean())

# Highest rated movie
print("\nHighest Rated Movie:")
print(df.loc[df["IMDb Rating"].idxmax(), ["Movie Name", "IMDb Rating"]])

# Lowest rated movie
print("\nLowest Rated Movie:")
print(df.loc[df["IMDb Rating"].idxmin(), ["Movie Name", "IMDb Rating"]])

# Top 5 movies
print("\nTop 5 Movies:")
print(df.nlargest(5, "IMDb Rating")[["Movie Name", "IMDb Rating"]])

# Search movie
name = input("\nEnter movie name to search: ")

print("\nMovie Details:")
print(df[df["Movie Name"].str.contains(name, case=False, na=False)])
```
